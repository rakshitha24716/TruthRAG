```python
def ask_rag(question):

    results = collection.query(
        query_texts=[question],
        n_results=3
    )

    context = ""

    for i, chunk in enumerate(results["documents"][0]):
        context += f"\n\n[Chunk {i+1}]\n{chunk}"

    prompt = f"""
    Answer ONLY from the provided context.

    Context:
    {context}

    Question:
    {question}

    Return format:

    Answer:
    <answer>

    Sources:
    - Chunk X
    """

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role":"user",
                "content":prompt
            }
        ],
        temperature=0
    )

    return response.choices[0].message.content
```

```python
from sentence_transformers import SentenceTransformer

similarity_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Model Loaded ✅")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def hallucination_score(answer, context):

    answer_embedding = similarity_model.encode(
        [answer]
    )

    context_embedding = similarity_model.encode(
        [context]
    )

    score = cosine_similarity(
        answer_embedding,
        context_embedding
    )[0][0]

    return score
```

```python
question = "What is the Two Pointer Technique?"

results = collection.query(
    query_texts=[question],
    n_results=3
)

context = "\n".join(
    results["documents"][0]
)
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role":"user",
            "content":f"""
            Answer ONLY from context:

            {context}

            Question:
            {question}
            """
        }
    ]
)

answer = response.choices[0].message.content

print(answer)
```

```python
score = hallucination_score(
    answer,
    context
)

print("Score:", score)
```

```python
if score > 0.75:
    print("✅ Supported by Context")
else:
    print("⚠️ Possible Hallucination")
```
